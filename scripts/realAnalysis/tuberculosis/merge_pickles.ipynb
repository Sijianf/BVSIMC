```python
import gzip, pickle, glob, os

PATH_ROOT = "/Users/sijianfan/projects/BiSSGL"  # <-- change to local root
PATH_OUTPUT = os.path.join(
    PATH_ROOT, "outputs/results/realAnalysis/tuberculosis/fold_cv"
)
PATH_ARCHIVE = os.path.join(PATH_OUTPUT, "archived")
```

```python
############### DRIMC ##############
files = sorted(glob.glob(f"{PATH_OUTPUT}/results_drimc_*.gz"))

display(files)

all_results = []
for f in files:
    with gzip.open(f, "rb") as fin:
        all_results.extend(pickle.load(fin))

with gzip.open(
    f"{PATH_OUTPUT}/results_drimc.gz",
    "wb",
) as fout:
    pickle.dump(all_results, fout)
```
