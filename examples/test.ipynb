```python
import bvsimc
print(bvsimc.__version__)

from bvsimc import BVSIMC
import numpy as np

np.random.seed(0)
Y = np.random.binomial(1, 0.3, size=(10, 8)).astype(float)
U = np.random.normal(size=(10, 4))
V = np.random.normal(size=(8, 4))

model = BVSIMC(Y, U, V, K=3, max_iter=20)
mu, A, B, logLik = model.optimization(seed=1)
print("A shape:", A.shape, "B shape:", B.shape)
```
