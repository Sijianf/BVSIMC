### Tuberculosis 

#### 1. Dataset

- Overall target: 

    - 4734 M. tuberculosis strains to 13 anti-tuberculosis drugs, namely **4734 strains vs. 13 drugs**. 

    - Strains: 355,709 binary features about gene mutation. 

    - Drugs: 28 binary features about chemical properties. 

- M2013: 
    
    - 47 strains, 39 genomic regions.

- T2015:
    
    - 7 drugs: First-line drugs: Isoniazid (INH), Rifampicin (RIF), Ethambutol (EMB), Pyrazinamide (PZA). Second-line drugs: Streptomycin (STM), Ofloxacin (OFL), Amikacin (AMK).

- L2016:

    - 7 drugs: First-line drugs: Isoniazid (INH), Rifampicin (RIF), Ethambutol (EMB), Pyrazinamide (PZA). Second-line drugs: Streptomycin (STM), Fluoroquinolones, Aminoglycosides. 


- F2018:

    - 6465 strains.

    - 14 drugs: Isoniazid (INH), Rifampicin (RIF), Ethionamide (ETH), Pyrazinamide (PZA), Ethambutol (EMB), Streptomycin (STM), Amikacin (AMK), Capreomycin (CAP), Kanamycin (KAN), Ciprofloxacin (CIP), Ofloxacin (OFL), Moxifloxacin (MOX), cycloserine (CYS) and para-aminosalicylic acid (PAS).

- Pre-screening: 

    - We only kept SNPs related with drug resistance according to WHO document from 2023 (updated in 2024), and excluded SNPs if not appeared in all considered strains. 

    - Further modifications can be made to reduce correlations among SNPs. Since we noticed several SNPs mapped to the same type of variant or mutation records in the WHO document, however, the exact nucleotide changes are not the same, which then caused a different naming when finding SNPs. 

- Further steps and challenages:

    - Not run on computer yet. Maybe it requires a long time to run. 

    - Prepare the similarity matrix for DRIMC and NRLMF. 

#### 2. Implementation

- The current dimention: 

    $\mathbf{Y} \in \mathbb{R}^{6949 \times 13}$, with row side information $\mathbf{U} \in \mathbb{R}^{6949 \times 9684}$ and column side information $\mathbf{V} \in \mathbb{R}^{13 \times 33}$.

    $\mathbf{A} \in \mathbb{R}^{9684 \times K}$, and $\mathbf{B} \in \mathbb{R}^{33 \times K}$.

    $\mathbf{Y} = \operatorname{expit}(\mathbf{UA}\mathbf{B}^\top\mathbf{V}^\top)$
